# v1 exploration — fibrosis (Tsukui, E-CURD-126)

A scratch notebook, not a writeup. Load the annotated object, check that the composition and the
UMAP look like lung rather than like noise, and confirm the CTHRC1⁺ signal is actually present
before building anything on top of it. Run after `snakemake results/cluster/annotated.h5ad`.

The argument lives in `v1_writeup.ipynb`; this is just the sanity check that comes first.

```python
import scanpy as sc
sc.settings.verbosity = 1
adata = sc.read_h5ad('../results/cluster/annotated.h5ad')
adata
```

```python
# composition by condition
import pandas as pd
pd.crosstab(adata.obs['cell_type'], adata.obs['condition'], normalize='columns')
```

```python
sc.pl.umap(adata, color=['leiden', 'condition'])
```

```python
# the CTHRC1+ pathologic fibroblast signal + stiffening machinery
for g in ['CTHRC1', 'COL1A1', 'LOXL2']:
    if g in (adata.raw.var_names if adata.raw is not None else adata.var_names):
        sc.pl.umap(adata, color=g, use_raw=adata.raw is not None)
```
